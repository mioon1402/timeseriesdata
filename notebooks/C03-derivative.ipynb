**3-1. 평균 속도가 12로 다가간다**

```python
def s(t): return t**3

print(f"{'dt':>10} {'간 거리':>16} {'평균 속도':>16}")
for dt in [1, 0.1, 0.01, 0.001, 0.0001, 0.00001]:
    간거리 = s(2+dt) - s(2)
    print(f"{dt:>10} {간거리:>16.10f} {간거리/dt:>16.10f}")

print("\n→ 12 로 다가간다.  s'(2) = 12")
print("  하지만 dt = 0 은 넣을 수 없다:")
try:
    print((s(2+0) - s(2)) / 0)
except ZeroDivisionError as e:
    print("  ZeroDivisionError —", e)
```

**3-2. 12 + 6dt + dt² 와 정확히 같은가**

```python
print(f"{'dt':>8} {'실제 평균 속도':>20} {'12 + 6dt + dt²':>20} {'차':>12}")
for dt in [1, 0.1, 0.01, 0.001]:
    실제 = (s(2+dt) - s(2)) / dt
    공식 = 12 + 6*dt + dt**2
    print(f"{dt:>8} {실제:>20.12f} {공식:>20.12f} {실제-공식:>12.2e}")

print("\n→ 완전히 같다. (2+dt)³ = 8 + 12dt + 6dt² + dt³ 를 전개한 결과다.")
print("  dt 로 나누면 12 + 6dt + dt² 이고, 이 식은 dt=0 에서도 멀쩡히 12 다.")
```

**3-3. 도함수 s'(t) = 3t²**

```python
import numpy as np

def 수치미분(f, x, h=1e-6):
    return (f(x+h) - f(x-h)) / (2*h)      # 중앙차분 — 한쪽 차분보다 훨씬 정확하다

print(f"{'t':>5} {'수치미분':>16} {'3t²':>12} {'차':>12}")
for t in [0.5, 1, 2, 3, 5]:
    print(f"{t:>5} {수치미분(s, t):>16.9f} {3*t**2:>12.6f} {수치미분(s,t)-3*t**2:>12.2e}")

print("\n→ 어느 t 에서도 3t² 과 일치한다. 도함수는 '기울기 지도' 다.")
```

**3-4. h를 너무 줄이면 벌어지는 일**

```python
참값 = 12.0
print(f"{'h':>10} {'중앙차분':>20} {'오차':>14}")
for k in range(1, 17):
    h = 10.0**(-k)
    값 = (s(2+h) - s(2-h)) / (2*h)
    print(f"{h:>10.0e} {값:>20.12f} {abs(값-참값):>14.2e}")

print("\n→ h = 1e-5 ~ 1e-6 근처가 가장 정확하고, 더 줄이면 오히려 나빠진다.")
print("  수학의 극한과 컴퓨터의 뺄셈은 다르다.")
```

**3-5. 위치 · 속도 · 가속도**

```python
import numpy as np

def d(f, h=1e-5):
    return lambda x: (f(x+h) - f(x-h)) / (2*h)

운동 = {
    "등속 3t":       lambda t: 3*t,
    "등가속 t²":     lambda t: t**2,
    "점점 빨라짐 t³/3": lambda t: t**3/3,
    "진동 2sin t":   lambda t: 2*np.sin(t),
}

t = 1.0
print(f"t = {t} 일 때\n")
print(f"{'운동':>18} {'위치':>10} {'속도':>10} {'가속도':>10}")
for 이름, f in 운동.items():
    v, a = d(f), d(d(f, 1e-4), 1e-4)
    print(f"{이름:>18} {f(t):>10.4f} {v(t):>10.4f} {a(t):>10.4f}")

print("\n→ 등속은 가속도 0. 진동은 가속도가 위치와 반대 부호 (s'' = -s).")
```
