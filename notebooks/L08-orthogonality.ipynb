**8-1. 내적, 길이, 각도**

```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)

u = np.array([3., 4.])
v = np.array([-4., 3.])
w = np.array([1., 2.])

print("u · v =", u @ v, "   ← 0 이면 수직")
print("u · w =", u @ w)
print()
print("‖u‖ =", np.linalg.norm(u), " = √(u·u) =", np.sqrt(u @ u))
print()

def 각도(a, b):
    cos = (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))
    cos = np.clip(cos, -1, 1)          # 반올림 오차로 1.0000001 이 되는 걸 막는다
    return np.degrees(np.arccos(cos))

print(f"u 와 v 사이 각 = {각도(u, v):.1f}°")
print(f"u 와 w 사이 각 = {각도(u, w):.1f}°")
```

**8-2. 피타고라스 정리 확인**

```python
print("u ⟂ v 인가:", np.isclose(u @ v, 0))
print()
print("‖u+v‖² =", np.linalg.norm(u + v) ** 2)
print("‖u‖² + ‖v‖² =", u @ u + v @ v)
print("같은가:", np.isclose(np.linalg.norm(u + v) ** 2, u @ u + v @ v))
print()

# 직교가 아니면 깨진다
print("u 와 w (수직 아님):")
print("  ‖u+w‖² =", np.linalg.norm(u + w) ** 2)
print("  ‖u‖² + ‖w‖² =", u @ u + w @ w)
print("  차이 =", np.linalg.norm(u + w) ** 2 - (u @ u + w @ w), " = 2·(u·w) =", 2 * (u @ w))
```

**8-3. 코사인 유사도 — 길이를 무시하고 방향만**

```python
문서A = np.array([3., 1., 0., 2.])      # 단어 빈도라고 하자
문서B = np.array([6., 2., 0., 4.])      # A 와 같은 비율, 분량만 2배
문서C = np.array([0., 0., 5., 1.])      # 전혀 다른 주제

def 코사인유사도(a, b):
    return (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("A vs B :", round(코사인유사도(문서A, 문서B), 4), " ← 분량만 다르고 방향이 같다")
print("A vs C :", round(코사인유사도(문서A, 문서C), 4))
print()
print("유클리드 거리로 재면:")
print("  A-B :", round(np.linalg.norm(문서A - 문서B), 3), " ← 분량 차이 때문에 멀어 보인다")
print("  A-C :", round(np.linalg.norm(문서A - 문서C), 3))
print()
print("→ '무엇에 대한 글인가' 를 보고 싶으면 코사인, '얼마나 많은가' 도 보려면 거리")
```

**8-4. 네 부분공간의 직교성 직접 확인**

```python
from scipy.linalg import null_space

A = np.array([[1., 2., 2., 2.],
              [2., 4., 6., 8.],
              [3., 6., 8., 10.]])

N  = null_space(A)      # 영공간   (ℝ⁴ 안)
LN = null_space(A.T)    # 좌영공간 (ℝ³ 안)

print("① 행공간 ⟂ 영공간")
print("   A @ N 최대 오차 :", np.abs(A @ N).max())
print("   → A 의 각 행이 영공간의 모든 벡터와 직교")
print()
print("② 열공간 ⟂ 좌영공간")
print("   A.T @ LN 최대 오차 :", np.abs(A.T @ LN).max())
print()

# 행공간의 '아무' 벡터를 만들어 확인 (행들의 선형결합)
rng = np.random.default_rng(0)
for _ in range(3):
    행공간_v = A.T @ rng.normal(size=3)
    print("   행공간 벡터 · 영공간 =", np.round(행공간_v @ N, 12))
```

**8-5. 직교 부분공간의 흔한 오해 확인**

```python
# 3차원에서 'xy 평면' 과 'xz 평면' 은 직교 부분공간이 아니다
xy_기저 = np.array([[1., 0., 0.], [0., 1., 0.]])
xz_기저 = np.array([[1., 0., 0.], [0., 0., 1.]])

print("두 평면의 모든 기저 쌍 내적:")
for i, a in enumerate(xy_기저):
    for j, b in enumerate(xz_기저):
        print(f"  xy[{i}] · xz[{j}] = {a @ b}")

print()
공유 = np.array([1., 0., 0.])
print("공유하는 벡터 (1,0,0) 는 양쪽 모두에 속한다.")
print("자기 자신과의 내적 =", 공유 @ 공유, " ≠ 0  →  직교 불가")
```

**모범 답안**

```python
# 문제 1
u1 = np.array([1., 2., 2.])
print("‖u‖ =", np.linalg.norm(u1), " (1+4+4=9 의 제곱근)")
단위 = u1 / np.linalg.norm(u1)
print("단위벡터 =", 단위, " 길이 =", np.linalg.norm(단위))

# 문제 2 — [1,1,0] 과 직교하는 것들은 2차원 평면을 이룬다
a = np.array([1., -1., 0.])
b = np.array([0., 0., 1.])
print("\n[1,1,0]·a =", np.array([1.,1.,0.]) @ a, "  ·b =", np.array([1.,1.,0.]) @ b)
print("두 벡터의 랭크 =", np.linalg.matrix_rank(np.column_stack([a, b])), "→ 2차원 평면")
print("(3차원에서 직선의 직교 여공간은 평면. 1 + 2 = 3)")

# 문제 3
p = np.array([1., 0.])
q = np.array([-0.5, np.sqrt(3)/2])          # 단위원 위 120°
cos = (p @ q) / (np.linalg.norm(p) * np.linalg.norm(q))
print("\ncos =", round(cos, 4), " 각도 =", round(np.degrees(np.arccos(np.clip(cos,-1,1))), 1), "°")
```
