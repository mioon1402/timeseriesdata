**13-1. 고윳값·고유벡터 구하기**

```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)

A = np.array([[2., 1.],
              [1., 2.]])

λ, V = np.linalg.eig(A)          # 고유벡터는 V 의 '열'
print("고윳값 λ =", λ)
print("\n고유벡터 (열마다 하나) =")
print(V)
print()

for i in range(len(λ)):
    x = V[:, i]
    print(f"λ={λ[i]:.4f}   A@x = {np.round(A @ x, 4)}   λ·x = {np.round(λ[i]*x, 4)}   같은가: {np.allclose(A @ x, λ[i]*x)}")
```

**13-2. 검산 — 대각합과 행렬식**

```python
print("trace(A) =", np.trace(A), "   λ 들의 합 =", λ.sum())
print("det(A)   =", round(np.linalg.det(A), 6), "   λ 들의 곱 =", round(λ.prod(), 6))
print()
print("특성방정식 λ² - (a+d)λ + (ad-bc) = 0 을 직접 풀어보면:")
tr, de = np.trace(A), np.linalg.det(A)
근 = np.roots([1, -tr, de])
print("  근 =", 근, "  ← eig 와 같다")
```

**13-3. 고유벡터 = 영공간**

```python
from scipy.linalg import null_space

for lam in λ:
    N = null_space(A - lam * np.eye(2))
    print(f"λ = {lam:.4f} 일 때 (A - λI) 의 영공간:")
    print(np.round(N, 4).ravel(), " ← 고유벡터")
    print()

print("→ 고유벡터를 구하는 건 6강의 '영공간 구하기' 와 똑같은 작업이다.")
```

**13-4. 네 가지 행렬 비교**

```python
행렬들 = {
    "대각 [[3,0],[0,1]]"  : np.array([[3., 0.], [0., 1.]]),
    "대칭 [[2,1],[1,2]]"  : np.array([[2., 1.], [1., 2.]]),
    "전단 [[1,1],[0,1]]"  : np.array([[1., 1.], [0., 1.]]),
    "회전 90°"            : np.array([[0., -1.], [1., 0.]]),
}

for 이름, Mx in 행렬들.items():
    w, v = np.linalg.eig(Mx)
    독립 = np.linalg.matrix_rank(v)
    복소 = np.iscomplexobj(w) and np.abs(w.imag).max() > 1e-12
    print(f"{이름:22s} λ = {np.round(w, 3)}")
    print(f"{'':22s} 독립인 고유벡터 {독립}개"
          + ("   ← 복소수! 실수 고유벡터 없음" if 복소 else "")
          + ("   ← 부족하다. 대각화 불가 (14강)" if (독립 < 2 and not 복소) else ""))
    print()
```

**13-5. 대칭행렬은 고유벡터가 직교한다**

```python
S = np.array([[4., 1., 0.],
              [1., 3., 1.],
              [0., 1., 2.]])                # 대칭

w, v = np.linalg.eigh(S)                    # 대칭 전용 함수 (더 빠르고 정확)
print("고윳값 =", w, "  ← 전부 실수")
print("\n고유벡터 =")
print(v)
print("\nVᵀV =")
print(np.round(v.T @ v, 10), "  ← 단위행렬 = 서로 직교하고 길이 1")
print()
print("→ 대칭행렬의 고유벡터는 자동으로 정규직교기저가 된다 (15강)")
```

**13-6. 고윳값 0 = 영공간**

```python
Z = np.array([[1., 2.],
              [2., 4.]])                    # 특이행렬

w, v = np.linalg.eig(Z)
print("고윳값 =", w, "  ← 하나가 0")
print("det =", round(np.linalg.det(Z), 10), " = λ 들의 곱")
print()
영_index = np.argmin(np.abs(w))
영벡터 = v[:, 영_index]
print("λ=0 의 고유벡터 =", np.round(영벡터, 4))
print("Z @ 그것 =", np.round(Z @ 영벡터, 10), "  ← 0. 바로 영공간이다")
```
