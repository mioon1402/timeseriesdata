## 1. 데이터 불러오기

**1-1. 데이터 열어보기**

```python
import pandas as pd

df = pd.read_csv("cafe_sales.csv")
df.head()
```

## 2. 얼마나 큰 데이터인가

**1-2. 크기 확인**

```python
print("행과 열:", df.shape)
print("행 개수:", len(df))
print("열 이름:", list(df.columns))
```

## 3. 어떤 열이 있고, 각 열은 무슨 타입인가

**1-3. 열 정보 훑어보기**

```python
df.info()
```

**1-4. 결측치 세어보기**

```python
df.isna().sum()
```

## 4. 숫자로 요약하기 — describe

**1-5. 기초 통계 한 방에**

```python
df.describe()
```

## 5. 요약값을 읽는 법

**1-6. 평균 vs 중앙값**

```python
mean = df["sales"].mean()
median = df["sales"].median()

print(f"평균   {mean:,.0f}원")
print(f"중앙값 {median:,.0f}원")
print(f"차이   {mean - median:,.0f}원  ({(mean/median - 1) * 100:+.1f}%)")
```

**1-7. 극단적인 날 찾아보기**

```python
# 방문객이 가장 많았던 날 3일
print("=== 최고 ===")
print(df.nlargest(3, "visitors")[["date", "weekday", "visitors", "sales"]])

# 가장 적었던 날 3일
print()
print("=== 최저 ===")
df.nsmallest(3, "visitors")[["date", "weekday", "visitors", "sales"]]
```
