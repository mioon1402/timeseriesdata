```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)

θ = np.pi / 6                       # 30도
변환 = {
    "회전 30°":  np.array([[np.cos(θ), -np.sin(θ)], [np.sin(θ), np.cos(θ)]]),
    "반사 x축":  np.array([[1., 0.], [0., -1.]]),
    "투영 x축":  np.array([[1., 0.], [0., 0.]]),
    "전단 k=1":  np.array([[1., 1.], [0., 1.]]),
    "확대 2배":  np.array([[2., 0.], [0., 2.]]),
    "찌그리기":  np.array([[2., 0.], [0., 0.5]]),
}
for 이름, A in 변환.items():
    λ = np.linalg.eigvals(A)
    d = np.linalg.det(A)
    복소 = "" if np.allclose(λ.imag, 0) else "  ← 실수 고유벡터 없음"
    print(f"{이름:9s} det={d:+6.3f}   λ={np.round(λ, 3)}{복소}")
```

```python
import numpy as np

A = np.array([[3., 1.],
              [1., 2.]])
e1, e2 = np.array([1., 0.]), np.array([0., 1.])

print("A e₁ =", A @ e1, "  ← A 의 1열")
print("A e₂ =", A @ e2, "  ← A 의 2열")
print()
x = np.array([2., 3.])
print("A x        =", A @ x)
print("2·1열+3·2열 =", 2 * A[:, 0] + 3 * A[:, 1], " ← 같다")
print("\n→ 기저벡터의 행선지만 알면 모든 점의 행선지가 정해진다.")
```

```python
import numpy as np

def 방향(A):
    d = np.linalg.det(A)
    if abs(d) < 1e-12: return "붕괴 (차원이 줄었다)"
    return "유지" if d > 0 else "뒤집힘 (거울상)"

for 이름, A in [("회전 90°", np.array([[0., -1.], [1., 0.]])),
                ("반사",     np.array([[0., 1.], [1., 0.]])),
                ("투영",     np.array([[1., 0.], [0., 0.]])),
                ("확대 3배", np.array([[3., 0.], [0., 3.]]))]:
    print(f"{이름:8s} det={np.linalg.det(A):+.2f}  넓이 {abs(np.linalg.det(A)):.2f}배  방향 {방향(A)}")
```

```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)

def 회전(도):
    t = np.radians(도)
    return np.array([[np.cos(t), -np.sin(t)], [np.sin(t), np.cos(t)]])

S = np.array([[1., 1.], [0., 1.]])      # 전단
R = 회전(30)

print("R S (전단 먼저, 그 다음 회전) =\n", R @ S)
print("S R (회전 먼저, 그 다음 전단) =\n", S @ R)
print("\n같은가:", np.allclose(R @ S, S @ R))
print("\n하지만 회전끼리는 순서가 상관없다:")
print("  R30 R45 == R75 :", np.allclose(회전(30) @ 회전(45), 회전(75)))
print("  R45 R30 == R75 :", np.allclose(회전(45) @ 회전(30), 회전(75)))
```

```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)

def 반사(도):                            # 원점을 지나고 각도 도° 인 직선에 대한 반사
    t = np.radians(2 * 도)
    return np.array([[np.cos(t), np.sin(t)], [np.sin(t), -np.cos(t)]])

F1, F2 = 반사(0), 반사(30)
print("det F1 =", np.linalg.det(F1), " det F2 =", np.linalg.det(F2))
print("\nF2 F1 =\n", F2 @ F1)
print("det =", round(float(np.linalg.det(F2 @ F1)), 4), " → +1, 회전이 되었다")

t = np.radians(60)
R60 = np.array([[np.cos(t), -np.sin(t)], [np.sin(t), np.cos(t)]])
print("\nR60 과 같은가:", np.allclose(F2 @ F1, R60))
print("→ 두 거울의 각도가 30° 면 회전각은 그 두 배인 60°.")
```

```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)

A = np.array([[2., 1.],
              [1., 2.]])
λ, B = np.linalg.eig(A)             # B 의 열이 고유벡터
print("고윳값 =", λ)
print("고유벡터(열) =\n", B)

D = np.linalg.inv(B) @ A @ B
print("\nB⁻¹AB =\n", D, " ← 완벽한 대각행렬")
print("\n닮은 행렬은 무엇이 같은가:")
print("  대각합  A:", round(float(np.trace(A)), 4), " D:", round(float(np.trace(D)), 4))
print("  행렬식  A:", round(float(np.linalg.det(A)), 4), " D:", round(float(np.linalg.det(D)), 4))
print("  고윳값  A:", np.sort(np.linalg.eigvals(A)), " D:", np.sort(np.linalg.eigvals(D)))
```

```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)

u = np.array([1., 2.])
u = u / np.linalg.norm(u)           # 반사 '거울면의 법선'
H = np.eye(2) - 2 * np.outer(u, u)

print("H =\n", H)
print("det H =", round(float(np.linalg.det(H)), 4), " → −1, 반사가 맞다")
print("H² = I 인가:", np.allclose(H @ H, np.eye(2)))
print()
print("법선 방향 u 는 뒤집힌다:  H u =", H @ u, " = −u:", np.allclose(H @ u, -u))
w = np.array([-2., 1.]) / np.sqrt(5)          # u 에 수직
print("거울면 위 w 는 그대로: H w =", H @ w, " = w:", np.allclose(H @ w, w))
print("\n→ 11강 QR 분해에서 쓴 그 반사가 이것이다.")
```

```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)

def 이동(dx, dy):
    return np.array([[1., 0., dx], [0., 1., dy], [0., 0., 1.]])
def 회전(도):
    t = np.radians(도)
    return np.array([[np.cos(t), -np.sin(t), 0.],
                     [np.sin(t),  np.cos(t), 0.],
                     [0., 0., 1.]])

점 = np.array([1., 0., 1.])          # (1, 0) 을 동차좌표로

print("이동만:", (이동(3, 2) @ 점)[:2])
print()
# '원점에서 90도 돌린 뒤 (3,2) 만큼 이동' — 하나의 행렬로 합친다
합성 = 이동(3, 2) @ 회전(90)
print("합성 행렬 =\n", 합성)
print("적용 결과:", (합성 @ 점)[:2])
print()
print("순서를 바꾸면 (먼저 이동, 그 다음 회전):",
      (회전(90) @ 이동(3, 2) @ 점)[:2])
print("→ 결과가 다르다. 그래픽스에서 순서를 틀리는 것이 가장 흔한 버그다.")
```

```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)

def 회전x(도):
    t = np.radians(도)
    return np.array([[1., 0., 0.],
                     [0., np.cos(t), -np.sin(t)],
                     [0., np.sin(t),  np.cos(t)]])
def 회전z(도):
    t = np.radians(도)
    return np.array([[np.cos(t), -np.sin(t), 0.],
                     [np.sin(t),  np.cos(t), 0.],
                     [0., 0., 1.]])

R = 회전z(40) @ 회전x(25)            # 두 회전을 섞으면? 여전히 회전이다
print("det R =", round(float(np.linalg.det(R)), 6), " (회전이면 +1)")
print("RᵀR = I 인가:", np.allclose(R.T @ R, np.eye(3)))

λ, V = np.linalg.eig(R)
i = int(np.argmin(np.abs(λ - 1)))     # λ = 1 인 고유벡터 = 회전축
축 = np.real(V[:, i]); 축 = 축 / np.linalg.norm(축)
print("\n고윳값 =", np.round(λ, 4))
print("회전축 =", np.round(축, 4))
print("정말 안 움직이는가:", np.allclose(R @ 축, 축))
```

```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)

# 문제 1 — 1열 (0,1): e₁ 이 위로 간다. 2열 (−1,0): e₂ 가 왼쪽으로.
#          → 반시계 90° 회전
A1 = np.array([[0., -1.], [1., 0.]])
print("문제 1: A e₁ =", A1 @ np.array([1., 0.]), " A e₂ =", A1 @ np.array([0., 1.]))
print("        → 반시계 90° 회전.  det =", round(float(np.linalg.det(A1)), 3),
      " λ =", np.round(np.linalg.eigvals(A1), 3), "(복소수 = 실수 고유벡터 없음)")
print("        네 번 적용하면 제자리:", np.allclose(np.linalg.matrix_power(A1, 4), np.eye(2)))

# 문제 2 — 넓이 2배 + 방향 뒤집기
A2 = np.array([[2., 0.], [0., -1.]])
print("\n문제 2: det =", np.linalg.det(A2),
      "→ ① 넓이가 2배가 되고 ② 방향이 뒤집힌다(거울상).")

# 문제 3 — 전단은 100번 해도 전단. k 만 100 이 된다
S = np.array([[1., 1.], [0., 1.]])
S100 = np.linalg.matrix_power(S, 100)
print("\n문제 3: S¹⁰⁰ =\n", S100)
print("        det =", round(float(np.linalg.det(S100)), 6), " ← 여전히 1, 넓이는 그대로")
print("        고윳값 =", np.linalg.eigvals(S100), " ← 여전히 1, 1")
print("        → 전단의 거듭제곱은 전단. k 만 100 배가 된다.")
print("          대각화가 안 되는 행렬이라 λ 만 봐서는 이 폭주를 알 수 없다(14강).")
```
