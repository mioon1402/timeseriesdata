## 1. 기본 통계량 총정리

**5-1. 하나씩 계산해보기**

```python
import pandas as pd
df = pd.read_csv("cafe_sales.csv", parse_dates=["date"])
v = df["visitors"]

print(f"개수    {v.count():>10,}")
print(f"합계    {v.sum():>10,}")
print(f"평균    {v.mean():>10.2f}")
print(f"중앙값  {v.median():>10.1f}")
print(f"최빈값  {v.mode()[0]:>10}")
print(f"분산    {v.var():>10,.1f}")
print(f"표준편차 {v.std():>9.2f}")
print(f"최솟값  {v.min():>10}")
print(f"최댓값  {v.max():>10}")
print(f"범위    {v.max() - v.min():>10}")
```

## 2. n으로 나눌까 n−1로 나눌까

**5-2. 같은 데이터, 다른 답**

```python
import numpy as np

print("pandas  v.std()          ", round(v.std(), 4), "  ← ddof=1 (표본표준편차)")
print("pandas  v.std(ddof=0)    ", round(v.std(ddof=0), 4), "  ← 모표준편차")
print()
print("numpy   np.std(v)        ", round(np.std(v), 4), "  ← ddof=0 이 기본!")
print("numpy   np.std(v, ddof=1)", round(np.std(v, ddof=1), 4))
```

## 3. 분위수 — 평균이 못 보는 것

**5-3. 분위수 살펴보기**

```python
분위 = v.quantile([0, 0.25, 0.5, 0.75, 0.90, 0.95, 1.0])
print(분위.round(1).to_string())

q1, q3 = v.quantile(0.25), v.quantile(0.75)
print()
print(f"IQR (가운데 50%의 폭) = {q3 - q1:.1f}명")
print(f"상위 5% 기준선 = {v.quantile(0.95):.0f}명")
```

## 4. 변동계수 — 단위가 다른 것끼리 비교

**5-4. 변동계수 비교**

```python
for col in ["visitors", "sales", "avg_temp"]:
    s = df[col].dropna()
    cv = s.std() / s.mean() * 100
    print(f"{col:10s} 평균 {s.mean():>10,.1f}  표준편차 {s.std():>9,.1f}  CV {cv:>6.1f}%")
```

## 5. 왜도와 첨도 — 모양을 숫자로

**5-5. 분포의 모양**

```python
for col in ["visitors", "sales", "avg_temp"]:
    s = df[col].dropna()
    print(f"{col:10s} 왜도 {s.skew():>7.3f}   첨도 {s.kurt():>7.3f}   "
          f"평균 {s.mean():>9,.1f}  중앙값 {s.median():>9,.1f}")
```

## 6. agg — 한 번에 여러 통계

**5-6. 여러 열 × 여러 통계**

```python
결과 = df.agg({
    "visitors": ["count", "mean", "median", "std", "min", "max"],
    "sales":    ["count", "mean", "median", "std", "min", "max"],
})
print(결과.round(1).to_string())
```

**5-7. describe 를 원하는 대로**

```python
# 기본 describe 는 25/50/75% 만 보여준다. 원하는 분위수를 지정할 수 있다.
df[["visitors", "sales"]].describe(percentiles=[0.05, 0.25, 0.5, 0.75, 0.95]).round(1)
```
