**9-1. 직선 위로 투영**

```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)

a = np.array([2., 1.])      # 방향
b = np.array([1., 3.])      # 목표

x_hat = (a @ b) / (a @ a)
p = x_hat * a               # 그림자
e = b - p                   # 오차

print("x̂ =", x_hat)
print("p  =", p)
print("e  =", e)
print()
print("a · e =", a @ e, "  ← 0 이면 수직으로 잘 내렸다")
print("‖b‖² =", b @ b, " = ‖p‖² + ‖e‖² =", p @ p + e @ e, " (피타고라스)")
```

**9-2. 정말 가장 가까운가**

```python
# 직선 위의 여러 점과 b 사이 거리를 재본다
print(f"{'배수 t':>8}  {'점 t·a':>16}  {'b 까지 거리':>12}")
for t in [0.0, 0.5, x_hat - 0.3, x_hat, x_hat + 0.3, 1.5, 2.0]:
    점 = t * a
    d = np.linalg.norm(b - 점)
    표시 = "  ← 최소!" if np.isclose(t, x_hat) else ""
    print(f"{t:8.2f}  {str(np.round(점, 2)):>16}  {d:12.4f}{표시}")
```

**9-3. 투영행렬 만들기**

```python
A = a.reshape(-1, 1)                      # 열이 하나인 행렬로
P = A @ np.linalg.inv(A.T @ A) @ A.T

print("P =")
print(P)
print()
print("P @ b =", P @ b, "  ← 위에서 구한 p 와 같다")
print()
print("P² = P 인가 :", np.allclose(P @ P, P))
print("Pᵀ = P 인가 :", np.allclose(P.T, P))
print("trace(P)   =", np.trace(P), " = 투영하는 공간의 차원")
```

**9-4. 3차원 → 평면으로 투영**

```python
A2 = np.array([[1., 0.],
               [0., 1.],
               [1., 1.]])                 # 두 열이 만드는 평면
b3 = np.array([1., 2., 6.])

P2 = A2 @ np.linalg.inv(A2.T @ A2) @ A2.T
p2 = P2 @ b3
e2 = b3 - p2

print("그림자 p =", p2)
print("오차   e =", e2)
print()
print("A2ᵀ @ e (모든 열과 직교해야) =", A2.T @ e2)
print("trace(P2) =", np.trace(P2), " = 평면의 차원 2 ✓")
print()
print("P2 의 고유값 =", np.round(np.linalg.eigvals(P2), 6), " ← 1이 2개, 0이 1개")
```

**9-5. I − P 도 투영행렬**

```python
I = np.eye(3)
Q = I - P2

print("Q = I - P 가 투영행렬인가")
print("  Q² = Q :", np.allclose(Q @ Q, Q))
print("  Qᵀ = Q :", np.allclose(Q.T, Q))
print("  trace(Q) =", np.trace(Q), " = 3 - 2 = 1 (직교 여공간의 차원)")
print()
print("b = Pb + Qb 로 쪼개지는가")
print("  Pb =", P2 @ b3)
print("  Qb =", Q @ b3)
print("  합 =", P2 @ b3 + Q @ b3, " = b ?", np.allclose(P2 @ b3 + Q @ b3, b3))
print("  두 조각이 직교? (Pb)·(Qb) =", round((P2 @ b3) @ (Q @ b3), 12))
```

**9-6. 정규직교 기저면 P = AAᵀ**

```python
# 직교하는 두 벡터를 단위벡터로 만들어 열로 세운다
q1 = np.array([1., 0., 1.]) / np.sqrt(2)
q2 = np.array([0., 1., 0.])
Q_mat = np.column_stack([q1, q2])

print("QᵀQ =")
print(Q_mat.T @ Q_mat, " ← 단위행렬")
print()
P_easy = Q_mat @ Q_mat.T                 # 역행렬 없이
P_full = Q_mat @ np.linalg.inv(Q_mat.T @ Q_mat) @ Q_mat.T
print("AAᵀ 와 전체 공식이 같은가:", np.allclose(P_easy, P_full))
```

**9-7. 연습문제**

```python
# 문제 1. b = [4, 0] 을 a = [1, 1] 방향으로 투영하면?
#         손으로 예상해보고 코드로 확인하세요.

# 문제 2. b 가 이미 직선 위에 있으면 (예: b = 3a) 투영 결과는?
#         오차는 얼마인가요?

# 문제 3. b 가 a 와 수직이면 투영 결과는?

# 아래에 직접 써보세요
```

**모범 답안**

```python
def 투영(a, b):
    x = (a @ b) / (a @ a)
    return x, x * a, b - x * a

# 문제 1 — 대각선 방향이므로 (2, 2)
a1 = np.array([1., 1.]); b1 = np.array([4., 0.])
x, p, e = 투영(a1, b1)
print(f"문제 1: x̂={x}, p={p}, e={e}, a·e={a1 @ e}")

# 문제 2 — 이미 직선 위면 그대로, 오차 0
b2 = 3 * a1
x, p, e = 투영(a1, b2)
print(f"문제 2: x̂={x}, p={p}, e={e}  ← 오차 0, Pb = b")

# 문제 3 — 수직이면 그림자가 원점
b3 = np.array([1., -1.])
x, p, e = 투영(a1, b3)
print(f"문제 3: x̂={x}, p={p}, e={e}  ← 그림자가 0. 고유값 0 인 경우")
```
