**5-1. 닫혀 있는지 직접 검사**

```python
import numpy as np

def 직선위인가(p, 기울기=2, 절편=0):
    return np.isclose(p[1], 기울기 * p[0] + 절편)

# ① 원점을 지나는 직선 y = 2x
u = np.array([1., 2.])
v = np.array([-3., -6.])
print("y = 2x 위의 두 점:", u, v)
print("합 =", u + v, " → 직선 위?", 직선위인가(u + v))
print("2배 =", 2 * u, " → 직선 위?", 직선위인가(2 * u))
print("영벡터 포함?", 직선위인가(np.zeros(2)))
```

**5-2. 원점을 안 지나면 무너진다**

```python
# ② y = 2x + 1 — 원점을 지나지 않는 직선
u2 = np.array([0., 1.])
v2 = np.array([1., 3.])
print("y = 2x + 1 위의 두 점:", u2, v2)
print("합 =", u2 + v2, " → 직선 위?", 직선위인가(u2 + v2, 절편=1))
print("영벡터 포함?", 직선위인가(np.zeros(2), 절편=1))
print()
print("→ 세 조건 중 두 개가 깨졌다. 부분공간이 아니다.")
```

**5-3. 1사분면은 왜 안 되나**

```python
def 제1사분면인가(p):
    return bool(np.all(p >= 0))

w = np.array([2., 3.])
print("w =", w, " 1사분면?", 제1사분면인가(w))
print("w + [1,1] =", w + np.array([1., 1.]), " 1사분면?", 제1사분면인가(w + np.array([1., 1.])))
print("(-1) × w  =", -1 * w, "  1사분면?", 제1사분면인가(-1 * w), " ← 여기서 탈락")
print()
print("덧셈에는 닫혀 있지만 '음수 배' 에서 밖으로 나간다.")
print("조건 세 개를 '전부' 만족해야 부분공간이다.")
```

**5-4. Span 은 항상 부분공간**

```python
rng = np.random.default_rng(0)
v1 = np.array([2., 1., 0.])
v2 = np.array([1., 3., 1.])

def span_점(c1, c2):
    return c1 * v1 + c2 * v2

# 무작위로 두 점을 뽑아 더해도 여전히 Span 안인가?
a = span_점(*rng.normal(size=2))
b = span_점(*rng.normal(size=2))

# 'Span 안' 인지는 v1, v2 로 표현되는지로 판정한다 (최소제곱 잔차가 0)
A = np.column_stack([v1, v2])
def span_안인가(p):
    해, 잔차, *_ = np.linalg.lstsq(A, p, rcond=None)
    return np.allclose(A @ 해, p)

print("a in Span :", span_안인가(a))
print("b in Span :", span_안인가(b))
print("a+b in Span :", span_안인가(a + b))
print("-3.7a in Span :", span_안인가(-3.7 * a))
print("영벡터 in Span :", span_안인가(np.zeros(3)))
```

**5-5. 행렬도 벡터다**

```python
# 2×2 행렬들의 집합도 벡터공간이다 — 더하고 상수배할 수 있으니까
M1 = np.array([[1., 2.], [3., 4.]])
M2 = np.array([[0., 1.], [1., 0.]])

print("M1 + M2 =\n", M1 + M2)
print("\n2.5 × M1 =\n", 2.5 * M1)
print("\n영벡터 역할:\n", np.zeros((2, 2)))
print()
print("2×2 행렬 공간의 차원은? → 자유롭게 정할 수 있는 칸이 4개이므로 4")
```

**모범 답안**

```python
# 문제 1 — 부분공간이다
a = np.array([3., -3.]); b = np.array([-1., 1.])
print("합의 합:", (a + b).sum(), " 상수배의 합:", (2.5 * a).sum(), " 영벡터:", np.zeros(2).sum())
print("→ 전부 0 이므로 부분공간 ✓  (원점을 지나는 직선이다)")

# 문제 2 — 부분공간이 아니다
c = np.array([1., 0.]); d = np.array([0., 1.])
print("\n합:", (c + d).sum(), "← 1이 아니라 2. 영벡터도 합이 0이라 불포함")
print("→ 부분공간 아님 ✗  (원점을 안 지나는 직선)")

# 문제 3 — 부분공간이 아니다
e = np.array([1., 0., 0.]); f = np.array([0., 1., 0.])
print("\n(1,0,0)+(0,1,0) =", e + f, " 0인 성분이 있나:", bool((e + f == 0).any()))
g = np.array([1., 1., 0.]); h = np.array([0., 0., 1.])
print("(1,1,0)+(0,0,1) =", g + h, " 0인 성분이 있나:", bool((g + h == 0).any()), "← 없다")
print("→ 덧셈에 닫혀 있지 않다. 부분공간 아님 ✗")
print("   (세 개의 평면을 합친 모양이라 '평평한 하나의 공간' 이 아니다)")
```
