**4-1. 정사각형의 세 조각**

```python
import numpy as np

x = 2.0
print(f"한 변 x = {x}\n")
print(f"{'dx':>7} {'두 띠 2x·dx':>14} {'모서리 dx²':>13} {'모서리 비중':>12} {'변화율':>11}")
for dx in [1, 0.5, 0.2, 0.05, 0.01]:
    띠 = 2*x*dx
    모서리 = dx*dx
    전체 = (x+dx)**2 - x**2
    print(f"{dx:>7} {띠:>14.5f} {모서리:>13.5f} {100*모서리/전체:>11.2f}% {전체/dx:>11.5f}")

print(f"\n2x = {2*x}   ← 변화율이 여기로 다가간다")
```

**4-2. 정육면체의 일곱 조각**

```python
x = 2.0
print(f"{'dx':>7} {'판 3개':>12} {'막대 3개':>12} {'꼭짓점':>12} {'판 비중':>10}")
for dx in [1, 0.5, 0.2, 0.05, 0.01]:
    판 = 3*x*x*dx
    막대 = 3*x*dx*dx
    꼭짓점 = dx**3
    전체 = (x+dx)**3 - x**3
    assert abs(판 + 막대 + 꼭짓점 - 전체) < 1e-12      # 세 조각의 합이 정확히 전체다
    print(f"{dx:>7} {판:>12.5f} {막대:>12.5f} {꼭짓점:>12.6f} {100*판/전체:>9.2f}%")

print(f"\n3x² = {3*x*x}   ← (부피 변화)/dx 가 여기로 다가간다")
```

**4-3. (xⁿ)′ = n·xⁿ⁻¹ 는 n이 뭐든 통한다**

```python
def 수치미분(f, x, h=1e-6):
    return (f(x+h) - f(x-h)) / (2*h)

x = 2.0
print(f"x = {x} 에서\n")
print(f"{'n':>8} {'수치미분':>16} {'n·xⁿ⁻¹':>16} {'차':>10}")
for n in [2, 3, 5, 1, 0, -1, -2, 0.5, 1/3, -0.5]:
    f = lambda t, n=n: t**n
    수치 = 수치미분(f, x)
    공식 = n * x**(n-1)
    print(f"{n:>8.4f} {수치:>16.9f} {공식:>16.9f} {수치-공식:>10.2e}")

print("\n→ 정수·음수·분수 전부 같은 공식이 통한다.")
```

**4-4. 넓이 고정 직사각형 — 1/x**

```python
x = 1.0
print("넓이를 1 로 유지하면서 가로를 dx 늘리면 세로는 얼마나 줄어야 하나\n")
print(f"{'dx':>7} {'붙는 넓이 y·dx':>16} {'실제 dy':>14} {'dy/dx':>12}")
for dx in [0.5, 0.2, 0.05, 0.01, 0.001]:
    y0 = 1/x
    y1 = 1/(x+dx)                 # 넓이 1 을 지키려면 세로는 이 값이어야 한다
    print(f"{dx:>7} {y0*dx:>16.6f} {y1-y0:>14.6f} {(y1-y0)/dx:>12.6f}")

print(f"\n-1/x² = {-1/x**2}   ← 여기로 다가간다")
```

**4-5. 넓이와 경계 — 원과 구**

```python
import numpy as np

def 수치미분(f, x, h=1e-6):
    return (f(x+h) - f(x-h)) / (2*h)

r = 2.0
쌍 = [
    ("원의 넓이 πr²",       lambda r: np.pi*r**2,      "원둘레 2πr",     lambda r: 2*np.pi*r),
    ("구의 부피 (4/3)πr³",  lambda r: 4/3*np.pi*r**3,  "겉넓이 4πr²",    lambda r: 4*np.pi*r**2),
    ("정사각형 넓이 x²",     lambda r: r**2,            "두 변 2x",       lambda r: 2*r),
    ("정육면체 부피 x³",     lambda r: r**3,            "세 면 3x²",      lambda r: 3*r**2),
]

for 이름, f, 경계이름, g in 쌍:
    print(f"{이름:<22} 미분 = {수치미분(f, r):>10.6f}   {경계이름:<14} = {g(r):>10.6f}")

print("\n→ 부피를 미분하면 언제나 '자라고 있는 경계'의 크기가 나온다.")
```
