**1-1. 고리로 쪼개서 더해보기**

```python
import numpy as np

R = 1.0

def 고리합(n, 어디="안쪽"):
    dr = R / n
    r = np.arange(n) * dr              # 각 고리의 안쪽 반지름
    if 어디 == "가운데": r = r + dr/2
    if 어디 == "바깥":   r = r + dr
    return (2*np.pi*r * dr).sum()      # Σ (둘레 × 두께)

print(f"{'n':>5} {'안쪽':>12} {'가운데':>12} {'바깥':>12}")
for n in [1, 2, 4, 8, 20, 80, 1000]:
    print(f"{n:>5} {고리합(n,'안쪽'):>12.6f} {고리합(n,'가운데'):>12.6f} {고리합(n,'바깥'):>12.6f}")

print(f"\n참값 πR² = {np.pi:.6f}")
```

**1-2. 오차가 정말 (n−1)/n 인가**

```python
print(f"{'n':>5} {'안쪽 합':>11} {'πR²(n-1)/n':>13} {'오차 %':>9}")
for n in [2, 5, 10, 50, 200]:
    s  = 고리합(n, "안쪽")
    예상 = np.pi * (n-1) / n
    print(f"{n:>5} {s:>11.6f} {예상:>13.6f} {100*(np.pi-s)/np.pi:>8.3f}%")

print("\n→ 정확히 일치한다. 오차는 πR²/n 이므로 n을 키우면 0으로 간다.")
```

**1-3. 넓이를 미분하면 둘레가 나온다**

```python
def A(r): return np.pi * r**2          # 넓이

r = 1.0
print(f"반지름 r = {r},  둘레 2πr = {2*np.pi*r:.8f}\n")
print(f"{'h':>10} {'(A(r+h)-A(r))/h':>18} {'차이':>12}")
for h in [0.5, 0.1, 0.01, 0.001, 0.0001]:
    변화율 = (A(r+h) - A(r)) / h
    print(f"{h:>10} {변화율:>18.8f} {변화율 - 2*np.pi*r:>12.8f}")

print("\n→ 차이가 정확히 πh 다.  π×0.0001 =", np.pi*0.0001)
```

**1-4. 같은 방법으로 구의 부피 만들기**

```python
# 고리 대신 '양파 껍질(구면)'로 쪼갠다.
# 반지름 r 인 구면의 겉넓이는 4πr² 이므로, 껍질 하나의 부피 ≈ 4πr²·dr

R = 1.0
for n in [10, 100, 10000]:
    dr = R / n
    r = (np.arange(n) + 0.5) * dr      # 가운데로 재기
    V = (4*np.pi*r**2 * dr).sum()
    print(f"n={n:>6}  Σ 4πr²·dr = {V:.8f}")

print(f"\n참값 (4/3)πR³ = {4/3*np.pi:.8f}")
print("→ 겉넓이를 적분하면 부피. 넓이·둘레와 완전히 같은 구조다.")
```
