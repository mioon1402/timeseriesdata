**9-1. 리만 합 — 세 가지 방식**

```python
import numpy as np

def 리만(f, a, b, n, 방식="left"):
    dx = (b - a) / n
    i = np.arange(n)
    x = a + (i + {"left": 0, "mid": 0.5, "right": 1}[방식]) * dx
    return np.sum(f(x) * dx)

f, a, b, 참값 = lambda x: x**2, 0, 1, 1/3

print(f"{'n':>6} {'왼쪽':>14} {'가운데':>14} {'오른쪽':>14}")
for n in [1, 6, 20, 100, 1000, 100000]:
    print(f"{n:>6}", "".join(f"{리만(f,a,b,n,m):>14.8f}" for m in ["left","mid","right"]))

print(f"\n참값 = {참값:.8f}")
print("→ 왼쪽은 모자라고 오른쪽은 넘친다. 가운데가 압도적으로 정확하다.")
```

**9-2. 오차가 줄어드는 속도**

```python
print(f"{'n':>8} {'왼쪽 오차':>14} {'가운데 오차':>16}")
for n in [10, 100, 1000, 10000]:
    왼 = 리만(f, a, b, n, "left") - 참값
    가 = 리만(f, a, b, n, "mid") - 참값
    print(f"{n:>8} {왼:>14.2e} {가:>16.2e}")

print("\n→ n 을 10배 하면 왼쪽 오차는 10배, 가운데 오차는 100배 줄어든다.")
print("  왼쪽은 O(1/n), 가운데는 O(1/n²). '두 번 작은 것' 이 여기서도 나온다(4강).")
```

**9-3. 면적 함수를 미분하면 원래 함수**

```python
import numpy as np

def 면적(f, a, x, n=20000):
    """a 부터 x 까지의 넓이 (가운데 리만합)"""
    if x == a: return 0.0
    dx = (x - a) / n
    t = a + (np.arange(n) + 0.5) * dx
    return np.sum(f(t) * dx)

g = lambda x: 1 + np.sin(x)          # 아무 함수나
a = 0.0

print(f"{'x':>6} {'A(x)':>14} {'A′(x) (수치)':>16} {'f(x)':>14}")
for x in [0.5, 1.0, 2.0, 3.0, 4.0]:
    A  = 면적(g, a, x)
    Ap = (면적(g, a, x+1e-5) - 면적(g, a, x-1e-5)) / 2e-5
    print(f"{x:>6} {A:>14.8f} {Ap:>16.8f} {g(x):>14.8f}")

print("\n→ A′(x) 와 f(x) 가 같다. 제1정리다.")
```

**9-4. 제2정리 — 양 끝값만 빼면 끝**

```python
import numpy as np

문제 = [
    ("∫₀¹ x² dx",      lambda x: x**2,     lambda x: x**3/3,        0, 1),
    ("∫₀² 3x² dx",     lambda x: 3*x**2,   lambda x: x**3,          0, 2),
    ("∫₀^π sin x dx",  np.sin,             lambda x: -np.cos(x),    0, np.pi),
    ("∫₁ᵉ (1/x) dx",   lambda x: 1/x,      np.log,                  1, np.e),
    ("∫₀¹ eˣ dx",      np.exp,             np.exp,                  0, 1),
]

print(f"{'식':>18} {'리만합 (n=100만)':>20} {'F(b) − F(a)':>18}")
for 이름, f, F, a, b in 문제:
    수치 = 리만(f, a, b, 1000000, "mid")
    공식 = F(b) - F(a)
    print(f"{이름:>18} {수치:>20.10f} {공식:>18.10f}")

print("\n→ 완전히 같다. 무한 합을 뺄셈 하나로 대신했다.")
```

**9-5. 역도함수가 없어도 넓이는 있다**

```python
import numpy as np

# e^(-x²) 는 초등함수로 된 역도함수가 존재하지 않는다.
# 그래도 넓이는 멀쩡히 있고, 정의(무한 합)대로 계산할 수 있다.
f = lambda x: np.exp(-x**2)

print("∫ e^(-x²) dx  (정규분포의 심장)")
for b in [1, 2, 3, 5, 10]:
    print(f"  -{b} ~ {b} : {리만(f, -b, b, 2000000, 'mid'):.10f}")

print(f"\n  전체 (-∞~∞) 의 참값 √π = {np.sqrt(np.pi):.10f}")
print("\n→ '거꾸로 미분' 이 불가능해도 넓이는 계산된다.")
print("  적분의 정의가 넓이이기 때문이다. 이 구분이 실무에서 결정적이다.")
```

**9-6. scipy 로 확인**

```python
from scipy.integrate import quad
import numpy as np

for 이름, f, a, b in [
    ("∫₀¹ x²",        lambda x: x**2,      0, 1),
    ("∫₀^π sin x",    np.sin,              0, np.pi),
    ("∫₋∞^∞ e^(-x²)", lambda x: np.exp(-x**2), -np.inf, np.inf),
]:
    값, 오차 = quad(f, a, b)
    print(f"{이름:>18} = {값:.12f}   (추정 오차 {오차:.1e})")

print("\n→ quad 는 적응형 구적법이다. 리만합보다 훨씬 적은 계산으로 훨씬 정확하다.")
print("  하지만 원리는 같다 — '높이 × 폭을 잘 골라 더한다'.")
```
