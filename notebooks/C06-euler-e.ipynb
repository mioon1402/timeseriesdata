**6-1. x=0 에서의 기울기 c 찾기**

```python
import numpy as np

def c_of(a, h=1e-7):
    """a^x 의 x=0 에서의 접선 기울기"""
    return (a**h - a**(-h)) / (2*h)

print(f"{'밑 a':>10} {'c (x=0 기울기)':>18} {'ln a':>14}")
for a in [1.5, 2, 2.5, np.e, 3, 4, 10]:
    print(f"{a:>10.5f} {c_of(a):>18.8f} {np.log(a):>14.8f}")

print("\n→ c 와 ln a 가 완전히 같다. c = 1 이 되는 a 가 e = 2.718281828… 이다.")
```

**6-2. c = 1 인 밑을 직접 찾아보기**

```python
# 이분법 — c(a) 는 a 에 대해 증가하므로 1을 사이에 두고 좁혀 나간다
lo, hi = 2.0, 3.0
for step in range(60):
    mid = (lo + hi) / 2
    if c_of(mid) < 1: lo = mid
    else:             hi = mid

print(f"c = 1 이 되는 밑 = {mid:.12f}")
print(f"참값 e         = {np.e:.12f}")
print(f"차             = {abs(mid - np.e):.2e}")
print("\n→ '기울기가 1이 되는 밑' 이라는 조건 하나로 e 가 결정된다.")
```

**6-3. 높이 = 기울기**

```python
def 수치미분(f, x, h=1e-6):
    return (f(x+h) - f(x-h)) / (2*h)

print(f"{'x':>6} {'eˣ (높이)':>14} {'기울기':>14} {'기울기÷높이':>14}")
for x in [-2, -1, 0, 0.5, 1, 2, 3]:
    높이 = np.exp(x)
    기울기 = 수치미분(np.exp, x)
    print(f"{x:>6} {높이:>14.6f} {기울기:>14.6f} {기울기/높이:>14.9f}")

print("\n→ 어디서든 비가 정확히 1. eˣ 는 '자기 자신을 도함수로 갖는' 유일한 꼴이다.")
print("\n2ˣ 와 3ˣ 도 해보면:")
for a in [2, 3]:
    f = lambda x, a=a: a**x
    비 = [수치미분(f, x)/f(x) for x in [-1, 0, 1, 2]]
    print(f"  a={a}: 기울기÷높이 = {[round(v,6) for v in 비]}   ln {a} = {np.log(a):.6f}")
```

**6-4. 복리 — 무한히 나눠도 e 를 못 넘는다**

```python
print(f"{'나누는 횟수 n':>16} {'(1+1/n)ⁿ':>18} {'e 와의 차':>14}")
for n in [1, 2, 4, 12, 52, 365, 8760, 10**6, 10**8]:
    v = (1 + 1/n)**n
    이름 = {1:"연 1회", 2:"반년", 4:"분기", 12:"매월", 52:"매주", 365:"매일", 8760:"매시간"}.get(n, "")
    print(f"{n:>16,} {v:>18.10f} {np.e - v:>14.2e}  {이름}")

print(f"\ne = {np.e:.10f}")
print("→ 무한히 잘게 나눠도 2.71828… 을 넘지 못한다.")
```

**6-5. 성장과 감쇠 — dy/dx = ky**

```python
import numpy as np

사례 = [
    ("연 3% 복리 저축",     0.03,  100,  "만원"),
    ("세균 (20분마다 2배)", np.log(2)/(20/60), 1, "마리"),
    ("탄소-14 (반감기 5730년)", -np.log(2)/5730, 100, "%"),
    ("커피 식기 (τ=15분)",  -1/15, 80,   "℃ 초과분"),
]

for 이름, k, y0, 단위 in 사례:
    print(f"{이름}   (k = {k:+.6f})")
    for t in [0, 1, 5, 10]:
        print(f"    t={t:>3} : {y0*np.exp(k*t):>12.5f} {단위}")
    print()

print("→ k>0 이면 폭발적 성장, k<0 이면 지수적 감쇠. 둘 다 같은 식이다.")
```
