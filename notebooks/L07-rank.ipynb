**7-1. 독립인지 판정하기**

```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)

def 판정(벡터들, 이름=""):
    A = np.column_stack(벡터들)          # 벡터를 열로 세운다
    r = np.linalg.matrix_rank(A)
    n = A.shape[1]
    print(f"{이름:22s} 벡터 {n}개, 랭크 {r}  →  "
          f"{'독립' if r == n else '종속 (' + str(n - r) + '개가 남아돈다)'}")

판정([[1, 0], [0, 1]],           "(1,0), (0,1)")
판정([[1, 2], [2, 4]],           "(1,2), (2,4)")           # 평행
판정([[1, 0], [0, 1], [3, 5]],   "2차원에 벡터 3개")
판정([[1, 0, 0], [0, 1, 0], [0, 0, 1]], "3차원 표준기저")
```

**7-2. 행 랭크 = 열 랭크**

```python
rng = np.random.default_rng(0)

for m, n in [(3, 5), (5, 3), (4, 4), (2, 7)]:
    A = rng.integers(-3, 4, size=(m, n)).astype(float)
    print(f"{m}×{n} 행렬:  rank(A) = {np.linalg.matrix_rank(A)}, "
          f"rank(Aᵀ) = {np.linalg.matrix_rank(A.T)}")

print("\n→ 모양이 어떻든 항상 같다. 이것이 '행 랭크 = 열 랭크' 정리다.")
print("   그래서 랭크는 min(m, n) 을 넘을 수 없다.")
```

**7-3. 네 부분공간의 차원**

```python
from scipy.linalg import null_space

A = np.array([[1., 2., 2., 2.],
              [2., 4., 6., 8.],
              [3., 6., 8., 10.]])
m, n = A.shape
r = np.linalg.matrix_rank(A)

print(f"A 는 {m}×{n} 행렬,  랭크 r = {r}\n")
print(f"  행공간   dim = r     = {r}       (ℝ^{n} 안)")
print(f"  영공간   dim = n - r = {n - r}       (ℝ^{n} 안)")
print(f"  {'':30s}합 = {n} = n  ✓")
print()
print(f"  열공간   dim = r     = {r}       (ℝ^{m} 안)")
print(f"  좌영공간 dim = m - r = {m - r}       (ℝ^{m} 안)")
print(f"  {'':30s}합 = {m} = m  ✓")

print("\n실제로 세어보면:")
print("  null_space(A)   모양:", null_space(A).shape,   "→ 영공간 차원", null_space(A).shape[1])
print("  null_space(A.T) 모양:", null_space(A.T).shape, "→ 좌영공간 차원", null_space(A.T).shape[1])
```

**7-4. 직교성 미리 확인 (8강 예고)**

```python
N  = null_space(A)      # 영공간   기저
LN = null_space(A.T)    # 좌영공간 기저

# 행공간의 벡터는 A 의 행들의 선형결합
행공간_벡터 = A.T @ np.array([1.5, -2.0, 0.7])

print("행공간 벡터 · 영공간 벡터 =", np.round(행공간_벡터 @ N, 12))
print("열공간 벡터 · 좌영공간 벡터 =", np.round((A @ np.array([1., 2., -1., 0.5])) @ LN, 12))
print()
print("→ 전부 0. 네 공간은 짝을 지어 수직이다. 8강에서 증명합니다.")
```

**7-5. 랭크가 알려주는 것**

```python
def 진단(A):
    m, n = A.shape
    r = np.linalg.matrix_rank(A)
    print(f"{m}×{n}, rank={r}")
    print(f"  열이 독립 (영공간 = 0)      : {r == n}   → 해가 있다면 유일")
    print(f"  행이 독립 (좌영공간 = 0)    : {r == m}   → 어떤 b 든 해가 존재")
    print(f"  정사각 + 풀랭크 (가역)      : {m == n and r == n}")
    print()

진단(np.array([[1., 0.], [0., 1.], [1., 1.]]))     # 키가 큰 행렬
진단(np.array([[1., 2., 3.], [4., 5., 6.]]))       # 납작한 행렬
진단(np.array([[2., 1.], [1., 3.]]))               # 정사각 가역
```

```python
# 문제 1 — m=5, n=3, r=3
m1, n1, r1 = 5, 3, 3
print(f"행공간 {r1}, 영공간 {n1-r1}, 열공간 {r1}, 좌영공간 {m1-r1}")
print("→ 영공간이 0차원이므로 열이 독립. 해가 있으면 유일하다.")
print("   좌영공간이 2차원이므로 '풀리지 않는 b' 가 잔뜩 있다.")

# 문제 2 — 영행렬뿐
print("\nrank(영행렬) =", np.linalg.matrix_rank(np.zeros((3, 4))))
print("→ 0이 아닌 원소가 하나라도 있으면 랭크는 최소 1이다.")

# 문제 3 — 3행 = 1행 + 4행? 아니다. 5=1+4, 7=2+5, 9=3+6 → 3행 = 1행 + 2행
A3 = np.array([[1., 2., 3.], [4., 5., 6.], [5., 7., 9.]])
print("\n1행 + 2행 =", A3[0] + A3[1], "  3행 =", A3[2])
print("rank =", np.linalg.matrix_rank(A3), " (3이 아니라 2)")
print("det  =", round(np.linalg.det(A3), 10), " → 특이행렬")
```
