## 5. numpy 로 계산해보기

**1-1. 벡터 만들고 더하기**

```python
import numpy as np

u = np.array([3, 1])
v = np.array([1, 3])

print("u      =", u)
print("v      =", v)
print("u + v  =", u + v)
print("2u     =", 2 * u)
print("-1.5v  =", -1.5 * v)
```

**1-2. 리스트 vs 배열**

```python
리스트 = [3, 1]
배열 = np.array([3, 1])

print("리스트 + 리스트 :", 리스트 + [1, 3])
print("배열   + 배열   :", 배열 + np.array([1, 3]))
print()
print("리스트 * 2 :", 리스트 * 2)
print("배열   * 2 :", 배열 * 2)
```

**1-3. 선형결합 직접 만들기**

```python
# 가중치를 바꿔가며 어디로 가는지 본다
for c1, c2 in [(1, 1), (2, -1), (0.5, 0.5), (0, 0), (-1, 2)]:
    w = c1 * u + c2 * v
    print(f"{c1:>5} · u + {c2:>5} · v  =  {w}")
```

**1-4. 선형결합을 행렬 곱으로**

```python
# 재료 벡터들을 '열'로 세워 붙이면 행렬이 된다
A = np.column_stack([u, v])
print("A =")
print(A)
print()

c = np.array([2, -1])          # 가중치를 모은 벡터
print("A @ c      =", A @ c)   # @ 가 행렬 곱
print("2u - 1v    =", 2 * u - 1 * v)
print("같은가:", np.array_equal(A @ c, 2 * u - 1 * v))
```

**1-5. 표준기저 — 좌표란 결국 선형결합**

```python
p = np.array([4, -2, 7])
I = np.eye(3, dtype=int)       # 3차원 단위행렬 = 표준기저 e1, e2, e3

print("e1, e2, e3 =")
print(I)
print()

재구성 = p[0] * I[:, 0] + p[1] * I[:, 1] + p[2] * I[:, 2]
print("4·e1 + (-2)·e2 + 7·e3 =", 재구성)
print("원래 p                =", p)
```

**1-6. 그림으로 확인하기**

```python
import matplotlib.pyplot as plt

def 화살표(ax, v, 색, 이름, 시작=(0, 0)):
    ax.annotate("", xy=(시작[0] + v[0], 시작[1] + v[1]), xytext=시작,
                arrowprops=dict(arrowstyle="-|>", lw=2, color=색))
    ax.text(시작[0] + v[0] * 0.55 + 0.15, 시작[1] + v[1] * 0.55 + 0.15,
            이름, color=색, fontsize=12, fontweight="bold")

fig, ax = plt.subplots(figsize=(5.4, 5.4))
화살표(ax, u, "#2563eb", "u")
화살표(ax, v, "#16a34a", "v")
화살표(ax, v, "#16a34a", "", 시작=u)      # 머리-꼬리 잇기
화살표(ax, u + v, "#dc2626", "u+v")

ax.set_xlim(-1, 6); ax.set_ylim(-1, 6)
ax.set_aspect("equal")                    # 각도가 왜곡되면 안 된다
ax.axhline(0, lw=1, color="gray"); ax.axvline(0, lw=1, color="gray")
ax.set_title("벡터 덧셈 = 머리에 꼬리 잇기")
plt.show()
```

**모범 답안**

```python
# 문제 1 — 대칭이므로 c1 = c2 = 1.25
print("문제 1:", 1.25 * u + 1.25 * v)

# 문제 2 — u 와 v 는 평행하지 않으므로 c1=c2=0 뿐이다 (선형독립)
#          이것이 3강에서 배울 '영공간이 원점뿐' 이라는 말과 같은 뜻
print("문제 2:", 0 * u + 0 * v, "→ 자명한 해뿐")

# 문제 3 — v2 가 u 의 -1.5 배이므로, 둘을 섞으면 u 방향 직선만 나온다
v2 = np.array([-4.5, -1.5])
print("v2 는 u 의", v2[0] / u[0], "배 →", "평행" if np.isclose(v2[0]/u[0], v2[1]/u[1]) else "독립")
print("[1, 3] 은 이 직선 위에 있나?", np.isclose(1 / 3, 3 / 1))
print("→ 만들 수 없다. Span 이 직선으로 쪼그라들었기 때문")
```
