**10-1. 표본 평균이 연속 평균으로 수렴한다**

```python
import numpy as np

f, a, b = np.sin, 0, np.pi

print(f"{'표본 개수 n':>12} {'표본 평균':>16} {'2/π 와의 차':>16}")
for n in [2, 4, 10, 100, 10000]:
    x = a + (np.arange(n) + 0.5) * (b - a) / n
    평균 = f(x).mean()
    print(f"{n:>12} {평균:>16.10f} {평균 - 2/np.pi:>16.2e}")

print(f"\n연속 평균 = ∫sin / π = 2/π = {2/np.pi:.10f}")
print(f"(최소+최대)/2 = {(0+1)/2}   ← 완전히 다른 값이다")
```

**10-2. 왜 0.5 가 아닌가 — 머무는 시간**

```python
import numpy as np

# sin 값을 구간별로 나눠 '얼마나 오래 그 높이에 있는가' 를 세어본다
x = np.linspace(0, np.pi, 200001)
y = np.sin(x)

print(f"{'높이 구간':>14} {'머무는 시간 비율':>18}")
for lo in np.arange(0, 1.0, 0.2):
    비율 = np.mean((y >= lo) & (y < lo + 0.2))
    print(f"  {lo:.1f} ~ {lo+0.2:.1f} {'':>4} {100*비율:>14.2f}%")

print("\n→ 0.8~1.0 구간에 머무는 시간이 0.0~0.2 보다 훨씬 길다.")
print("  높은 곳에 오래 있으니 평균이 0.5 보다 위로 올라간다.")
```

**10-3. 평균값 정리 — 평균 기울기 = 할선 기울기**

```python
import numpy as np
from scipy.optimize import brentq

F  = lambda x: x**3/3 - x
f  = lambda x: x**2 - 1        # F′

a, b = 0.0, 2.0
할선 = (F(b) - F(a)) / (b - a)

# f 의 평균 = 넓이 / 길이 (가운데 리만합)
n = 2000000
t = a + (np.arange(n) + 0.5) * (b - a) / n
평균 = f(t).mean()

print(f"할선 기울기      = {할선:.10f}")
print(f"f 의 평균 높이   = {평균:.10f}")
print(f"차               = {할선 - 평균:.2e}\n")

c = brentq(lambda x: f(x) - 할선, a + 1e-9, b)
print(f"평균을 실현하는 점 c = {c:.10f}")
print(f"  F′(c) = {f(c):.10f}   ← 할선 기울기와 같다")
```

**10-4. 평균선 위·아래 넓이가 같다**

```python
import numpy as np

f, a, b = np.sin, 0, np.pi
n = 2000000
x = a + (np.arange(n) + 0.5) * (b - a) / n
dx = (b - a) / n
m = f(x).mean()

위 = np.sum(np.maximum(f(x) - m, 0) * dx)
아래 = np.sum(np.maximum(m - f(x), 0) * dx)

print(f"평균 높이 m = {m:.10f}")
print(f"평균선 위로 삐져나온 넓이 = {위:.10f}")
print(f"평균선 아래로 모자란 넓이 = {아래:.10f}")
print(f"차                        = {위 - 아래:.2e}")
print("\n→ 정확히 같다. '평균' 의 정의가 바로 이것이다.")
```

**10-5. 기본정리를 거꾸로 읽기**

```python
import numpy as np

사례 = [
    ("sin, [0, π]",      np.sin,           lambda x: -np.cos(x),  0, np.pi),
    ("x², [0, 2]",       lambda x: x**2,   lambda x: x**3/3,      0, 2),
    ("eˣ, [0, 1]",       np.exp,           np.exp,                0, 1),
    ("1/x, [1, e]",      lambda x: 1/x,    np.log,                1, np.e),
]

print(f"{'함수':>14} {'평균 높이':>14} {'(F(b)-F(a))/(b-a)':>22} {'넓이':>12}")
for 이름, f, F, a, b in 사례:
    n = 500000
    x = a + (np.arange(n) + 0.5) * (b - a) / n
    평균 = f(x).mean()
    공식 = (F(b) - F(a)) / (b - a)
    print(f"{이름:>14} {평균:>14.8f} {공식:>22.8f} {평균*(b-a):>12.8f}")

print("\n→ 넓이 = 평균 높이 × 구간 길이 = F(b) - F(a).  기본정리의 두 번째 증명이다.")
```

**10-6. 연습문제**

```python
# 문제 1. cos x 의 0~π/2 평균 높이를 구하세요. (역도함수는 sin x)

# 문제 2. x² 의 0~1 평균 높이는? 그리고 그 값을 실현하는 점 c 는 어디일까요?

# 문제 3. 속도 v(t) = 3t² 로 t=0~2 를 달렸습니다.
#         평균 속도는? 그리고 정확히 그 속도였던 순간은 언제일까요?

# 아래에 직접 써보세요
```

**모범 답안**

```python
import numpy as np

def 평균높이(f, a, b, n=1000000):
    x = a + (np.arange(n) + 0.5) * (b - a) / n
    return f(x).mean()

# 문제 1 — (sin(π/2) - sin 0)/(π/2) = 2/π
print(f"문제 1: 수치 {평균높이(np.cos, 0, np.pi/2):.8f}   공식 2/π = {2/np.pi:.8f}\n")

# 문제 2 — 1/3, c = 1/√3
m = 평균높이(lambda x: x**2, 0, 1)
print(f"문제 2: 평균 = {m:.8f} (= 1/3),  c = √(1/3) = {np.sqrt(1/3):.8f}")
print(f"        확인: c² = {(1/3)**0.5**2:.8f}\n")

# 문제 3 — 평균 속도 4, 그 순간 t = 2/√3
m3 = 평균높이(lambda t: 3*t**2, 0, 2)
print(f"문제 3: 평균 속도 = {m3:.8f} m/s  (= (2³-0)/2 = 4)")
print(f"        3t² = 4 → t = {np.sqrt(4/3):.8f} 초")
print("        구간단속이 잡아내는 것이 바로 이 순간이다.")
```
