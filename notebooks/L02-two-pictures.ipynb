**2-1. 연립방정식 풀기**

```python
import numpy as np

# 2x - y = 1
#  x + y = 5
A = np.array([[2, -1],
              [1,  1]])
b = np.array([1, 5])

x = np.linalg.solve(A, b)
print("해 (x, y) =", x)
```

**2-2. 두 관점이 같은지 확인**

```python
# 행 관점 — 각 식이 성립하는가
print("2x - y =", 2*x[0] - x[1], "  (목표 1)")
print(" x + y =",   x[0] + x[1], "  (목표 5)")
print()

# 열 관점 — 열벡터를 x배, y배 해서 더하면 b 가 되는가
열1 = A[:, 0]      # 첫 번째 열 = [2, 1]
열2 = A[:, 1]      # 두 번째 열 = [-1, 1]
print("열1 =", 열1, "  열2 =", 열2)
print("x·열1 + y·열2 =", x[0]*열1 + x[1]*열2)
print("b            =", b)
print()
print("A @ x =", A @ x, "  ← 같은 계산이다")
```

**2-3. 해가 없으면 어떻게 되나**

```python
# 두 열이 평행한 행렬 — 열2 는 열1 의 정확히 2배
S = np.array([[1, 2],
              [2, 4]])

print("행렬식 det =", np.linalg.det(S), " ← 0 이면 특이행렬")
print()

try:
    np.linalg.solve(S, np.array([3, 7]))
except np.linalg.LinAlgError as e:
    print("풀 수 없음:", e)
    print("→ 목표 [3, 7] 이 두 열의 Span(직선) 밖에 있다")
```

**2-4. 목표가 Span 안에 있으면**

```python
# [3, 6] 은 열1=[1,2] 의 3배 → 직선 위에 있다
b2 = np.array([3, 6])

print("A 의 랭크        :", np.linalg.matrix_rank(S))
print("A|b 를 붙인 랭크 :", np.linalg.matrix_rank(np.column_stack([S, b2])))
print("→ 두 랭크가 같으면 해가 존재한다 (지금은 무한히 많다)")
print()

# solve 는 여전히 못 쓴다. lstsq 가 해 하나를 골라준다
해, *_ = np.linalg.lstsq(S, b2, rcond=None)
print("해 하나:", 해, " → 확인:", S @ 해)
print()
print("다른 해도 있다:", np.array([3.0, 0.0]), "→ 확인:", S @ np.array([3.0, 0.0]))
```

**모범 답안**

```python
# 문제 1
A1 = np.array([[1, 1],
               [2, 3]])
b1 = np.array([5, 13])
해1 = np.linalg.solve(A1, b1)
print("문제 1 → 아메리카노", 해1[0], "잔, 라떼", 해1[1], "잔")

# 문제 2 — 열의 선형결합으로 확인
print("문제 2 →", 해1[0]*A1[:, 0] + 해1[1]*A1[:, 1], "= b1 ?", b1)

# 문제 3
해3 = np.linalg.solve(A1, np.array([5, 20]))
print("문제 3 →", 해3, "  (라떼 5잔, 아메리카노 0잔이면 15천원이라 20천원은 불가능)")
print("        수학적 해는 있지만 '잔 수가 음수' 라 현실에는 없는 답이다")
```
